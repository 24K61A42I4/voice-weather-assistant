```python
import requests

def get_weather(village):

    response = requests.get(
        "https://api.openweathermap.org/data/2.5/weather",
        params={
            "q": village + ",IN",
            "appid": API_KEY,
            "units": "metric"
        },
        timeout=15
    )

    print("Status code:", response.status_code)

    if response.ok:
        data = response.json()

        description = data["weather"][0]["description"]
        temperature = data["main"]["temp"]
        humidity = data["main"]["humidity"]
        wind_speed = data["wind"]["speed"]

        weather_report = (
            f"Weather report for {data['name']}. "
            f"Description: {description}. "
            f"Temperature: {temperature} degrees Celsius. "
            f"Humidity: {humidity} percent. "
            f"Wind speed: {wind_speed} meters per second."
        )

        return weather_report

    else:
        return "Error: Unable to find weather in that location."
```
